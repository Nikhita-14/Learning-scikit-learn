# Lesson 2) Label Encoding

## ~ Create a Dataset

```python
import pandas as pd
dataset = {
    'Color':['Red', 'Blue', 'Green', 'Blue','Red','Green','Blue','Red','Green'],
    'Size':['Small', 'Medium', 'Large', 'Small','Medium','Large','Small','Medium','Large'],
    'City':['Delhi','Mumbai','Bangalore','Chennai','Hyderabad','Delhi','Chennai','Hyderabad','Mumbai'],
    'Price':[120,150,80,200,130,450,232,700,211],
    'Purchased':['Yes','No','No','Yes','Yes','No','No','Yes','No']
}

df=pd.DataFrame(dataset)
df
```

## ~ Definition of Label Encoding (Simple):
<p style="font-size: 17px;">Machine doesn't want text. It wants numbers only. <br>
    Converting categories into numbers, where each unique category gets a number.<br>
    The assigning of numbers is done by sorting of categories alphabetically.<br>
    For eg) <br>
    Red, Blue, Green. Blue-0, Green-1, Red-2<br>
    Yes, No. No-0, Yes-1 </p>

```python
from sklearn.preprocessing import LabelEncoder
```

## ~ Practicing Label Encoding on Each Column

```python
label_enc=LabelEncoder()
```

```python
y=df['Purchased']
```

```python
y_enc=label_enc.fit_transform(y)
```

```python
y_enc
```

```python
z=df['Color']
```

```python
z_enc=label_enc.fit_transform(z)
```

```python
z_enc
```

```python
a=df['City']
```

```python
a_enc=label_enc.fit_transform(a)
```

```python
a_enc
```

```python
b=df['Size']
```

```python
b_enc=label_enc.fit_transform(b)
```

```python
b_enc
```

## ~ Replacing Categories with Encoded values in the columns (ML Ready Dataset)

```python
for col in ['Color','Size','City','Purchased']:
    le=LabelEncoder()
    df[col]=le.fit_transform(df[col])
```
