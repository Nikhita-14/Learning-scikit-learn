## ~ Import pandas to read csv and train_test_split from sklearn

```python
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
```

## ~ Read the csv

```python
df=pd.read_csv(r"C:\Users\study\Desktop\Learning scikit-learn\data\student_grades_50.csv")
df.head()
```

```python
df.shape
```

## ~ Determining X and y
<p><strong>X (features/inputs):</strong> The data used to make predictions. Example: hours studied, attendance, previous grades.</p>

<p><strong>y (target/output):</strong> The value we want the model to predict. Example: pass/fail or final grade.</p>

<p><strong>How to determine X and y from a dataset:</strong></p>
<ul>
  <li>y: Choose the column you want to predict.</li>
  <li>X: Choose all other columns that help predict y.</li>
</ul>

```python
X=df[['Hours_Studied','Attendance','Previous_Grades']]
X.head()
```

```python
X.shape
```

```python
y=df[['Pass']]
y.head()
```

```python
y.shape
```

## ~ Performing train_test_split
<p>This line splits the dataset into training and testing sets:</p>

<ul>
  <li><strong>X_train:</strong> Features used to <em>train the model</em> (80% of data).</li>
  <li><strong>X_test:</strong> Features used to <em>test the model</em> (20% of data).</li>
  <li><strong>y_train:</strong> Target values for training.</li>
  <li><strong>y_test:</strong> Target values for testing.</li>
</ul>

<p><strong>Parameters:</strong></p>
<ul>
  <li><code>test_size=0.2</code> → 20% of data for testing.</li>
  <li><code>random_state=14</code> → ensures the split is reproducible.</li>
</ul>

<p>This ensures the model is trained on one set of data and evaluated on a separate unseen set to check performance.</p>

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=14
)
```

<h4>Train-Test Split Explanation</h4>
<p>When we split the dataset using <code>train_test_split</code>:</p>
<ul>
  <li><strong>Shuffle:</strong> Rows are randomly mixed before splitting (default behavior) to avoid bias from the original order. Default is True but if you don't want then you keep it false.</li>
  <li><strong>Shape change:</strong> The original dataset is divided into training and testing sets, so the number of rows in <code>X_train</code> and <code>X_test</code> changes according to <code>test_size</code>, but the number of columns stays the same.</li>
</ul>

```python
print(X_train.head())
print("X_train Shape:", X_train.shape)
print(X_test.head())
print("X_test Shape:", X_test.shape)
print(y_train.head())
print("y_train Shape:", y_train.shape)
print(y_test.head())
print("y_test Shape:", y_test.shape)
```

<h4>Using <code>describe()</code> in Machine Learning</h4>

<p>The <code>describe()</code> method provides <strong>summary statistics</strong> for numeric columns in a dataset:</p>

<ul>
  <li><strong>count:</strong> Number of non-missing values</li>
  <li><strong>mean:</strong> Average value</li>
  <li><strong>std:</strong> Standard deviation (spread of data)</li>
  <li><strong>min / max:</strong> Smallest and largest values</li>
  <li><strong>25%, 50%, 75%:</strong> Percentiles (quartiles) showing data distribution</li>
</ul>

<p><strong>Why it’s useful for ML:</strong></p>
<ul>
  <li>Helps understand feature distributions before training a model</li>
  <li>Identifies outliers, ranges, and spread of data</li>
  <li>Aids in feature scaling and preprocessing decisions</li>
</ul>
